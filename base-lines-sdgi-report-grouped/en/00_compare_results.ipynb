# Compare SDGi Report-Grouped English-Only Baselines



## 1. Load the results


```python
import os
import pandas as pd
from IPython.display import display

RESULTS_DIR = "/Users/aman/Desktop/Thesis/base-lines-sdgi-report-grouped/en"
model_dirs = ['xlm_roberta', 'bert', 'scibert', 'biobert', 'roberta_base', 'deberta_v3']

rows = []
for name in model_dirs:
    path = f"{RESULTS_DIR}/{name}/results.csv"
    if os.path.isfile(path):
        rows.append(pd.read_csv(path))
    else:
        print("No results yet:", name)
```

## 2. Compare and save the results

Results are sorted by validation samples F1. Precision, recall and F1 are averaged across passages; accuracy requires an exact label-set match.

The all and en results are compared separately because their test sets differ.


```python
if rows:
    comparison = pd.concat(rows, ignore_index=True)
    columns = ["model", "validation_loss", "validation_f1", "test_loss",
               "test_precision", "test_recall", "test_f1",
               "test_accuracy"]
    comparison = comparison.sort_values("validation_f1", ascending=False)
    display(comparison[columns].round(4))
    comparison.to_csv(f"{RESULTS_DIR}/comparison.csv", index=False)
    print(f"Completed models: {len(comparison)} / {len(model_dirs)}")
    print("Saved:", f"{RESULTS_DIR}/comparison.csv")
else:
    print("Run the training notebooks first, then rerun this notebook.")
```
